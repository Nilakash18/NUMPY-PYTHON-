Array Stacking And splitting



```python
a= np.array([1,2,3])
b= np.array([4,5,6])
print(np.vstack((a,b)))# vertical staking  making 1 D Array TO 2 D Array
```

```python
a= np.array([1,2,3])
b= np.array([4,5,6])
print(np.hstack((a,b)))#horinzental stacking  making no changes to dimentional(D)
```

```python
c=np.array([[1,2,3],[4,5,6]])
print(c)
```

splitting

```python
print(np.hsplit(c,3)) # general
```

```python
split = np.hsplit(c,3)

for s in split:
  print(s)
```

```python
split = np.hsplit(c,3)
print(split[0])
```

```python
split = np.hsplit(c,3)
print(split[1])
```

```python
vsplit = np.vsplit(c,2)

for s in vsplit:
  print(s)
```
